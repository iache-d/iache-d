```python
import numpy as np
import io
import matplotlib.pyplot as plt


matriz_mo = np.atleast_2d(np.genfromtxt(io.StringIO(datos_mo_raw), skip_header=2))
matriz_si = np.atleast_2d(np.genfromtxt(io.StringIO(datos_si_raw), skip_header=2))

longitudes_onda = matriz_mo[:, 0]
delta_mo = matriz_mo[:, 1]
beta_mo = matriz_mo[:, 2]
delta_si = matriz_si[:, 1]
beta_si = matriz_si[:, 2]


d_mo = 2.79  
d_si = 4.18  
N = 40       
n_aire = 1.0 

reflectancia_R = []


for i in range(len(longitudes_onda)):
    lam = longitudes_onda[i]
    

    n_mo = (1 - delta_mo[i]) - 1j * beta_mo[i]
    n_si = (1 - delta_si[i]) - 1j * beta_si[i]
    n_sustrato = n_si 
    
    fase_mo = (2 * np.pi * n_mo * d_mo) / lam
    fase_si = (2 * np.pi * n_si * d_si) / lam
    
    # Matrices con convención +i
    M_Mo = np.array([
        [np.cos(fase_mo), (1j / n_mo) * np.sin(fase_mo)],
        [1j * n_mo * np.sin(fase_mo), np.cos(fase_mo)]
    ])
    
    M_Si = np.array([
        [np.cos(fase_si), (1j / n_si) * np.sin(fase_si)],
        [1j * n_si * np.sin(fase_si), np.cos(fase_si)]
    ])
    
    M_bicapa = np.matmul(M_Mo, M_Si)
    M_total = np.linalg.matrix_power(M_bicapa, N)
    
    m11, m12 = M_total[0, 0], M_total[0, 1]
    m21, m22 = M_total[1, 0], M_total[1, 1]
    
    num = n_aire * (m11 + n_sustrato * m12) - (m21 + n_sustrato * m22)
    den = n_aire * (m11 + n_sustrato * m12) + (m21 + n_sustrato * m22)
    r = num / den
    
    R = np.abs(r)**2
    reflectancia_R.append(R)

reflectancia_R = np.array(reflectancia_R)


idx_max = np.nanargmax(reflectancia_R)
R_max = reflectancia_R[idx_max]
lambda_max = longitudes_onda[idx_max]

mitad_R = R_max / 2.0
indices_sobre_mitad = np.where(reflectancia_R >= mitad_R)[0]
lambda_izq = longitudes_onda[indices_sobre_mitad[0]]
lambda_der = longitudes_onda[indices_sobre_mitad[-1]]
fwhm = lambda_der - lambda_izq

print(f"--- RESULTADOS DEL ESPEJO EUV ---")
print(f"Reflectancia Máxima (R_max): {R_max:.4f} ({R_max*100:.2f}%)")
print(f"Longitud de onda del pico (lambda_max): {lambda_max:.4f} nm")
print(f"Ancho de banda (FWHM): {fwhm:.4f} nm")

plt.figure(figsize=(10, 6))
plt.plot(longitudes_onda, reflectancia_R, color='navy', linewidth=2, label='R(λ) calculada')
plt.plot(lambda_max, R_max, 'ro', label=f'Máximo: R={R_max:.4f} en λ={lambda_max:.2f} nm')
plt.hlines(y=mitad_R, xmin=lambda_izq, xmax=lambda_der, color='green', linestyle='--', label=f'FWHM: {fwhm:.4f} nm')

plt.title('Reflectancia de Multicapa Mo/Si (40 bicapas) en rango EUV', fontsize=14)
# La 'r' antes de la comilla evita el SyntaxWarning
plt.xlabel(r'Longitud de Onda $\lambda$ (nm)', fontsize=12) 
plt.ylabel('Reflectancia $R$', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(loc='upper right', fontsize=11)

plt.tight_layout()
plt.show()
```
