### Pico de Bragg

El **pico de Bragg** es la forma característica de la curva de dosis en profundidad $D(z)$ de una partícula cargada pesada (como el protón) al atravesar un medio. A diferencia de los fotones, la dosis del protón es baja a la entrada (*plateau*), crece a medida que el protón penetra, alcanza un **máximo agudo y localizado** justo antes de detenerse (el pico), y luego cae casi a cero.

**Origen físico.** La dosis depositada por unidad de profundidad es proporcional a la *stopping power* $S = -dE/dx$, dada por la ecuación de Bethe–Bloch. El factor dominante es $S \propto 1/\beta^2 \propto 1/v^2$: a medida que el protón se frena, su velocidad disminuye y deposita **cada vez más energía por unidad de camino**. El proceso se retroalimenta (al frenarse pierde energía más rápido, lo que lo frena aún más), produciendo el máximo justo donde la partícula se detiene. Una vez detenida, no hay más partícula que deposite dosis, por lo que $D(z)$ cae abruptamente a cero más allá del rango $R_0$.

**Descripción cuantitativa.** Usando la relación rango–energía de Bragg–Kleeman,

$$R_0 = \alpha\, E_0^{\,p}, \qquad p \approx 1.77 \text{ (agua)},$$

se obtiene que la dosis en profundidad escala (modelo de Bortfeld) como

$$D(z) \propto (R_0 - z)^{\,\frac{1}{p}-1}, \qquad z < R_0,$$

con exponente $\tfrac{1}{p}-1 \approx -0.435$. Esto representa una singularidad integrable en $z=R_0$ que, en la práctica, no diverge: queda suavizada por el *energy straggling* (los protones no se detienen todos exactamente a la misma profundidad) y por la captura de electrones del protón a muy baja energía. El resultado es un pico de altura y anchura finitas.

**Relevancia clínica.** La profundidad del pico se controla con la energía inicial $E_0$, lo que permite ubicar la dosis máxima en el tumor minimizando el daño al tejido sano, especialmente al tejido distal (detrás del tumor), que recibe dosis casi nula. Como un único pico es muy estrecho, se superponen haces de distintas energías y pesos para formar un **pico de Bragg extendido (SOBP)**, una meseta de dosis uniforme que cubre toda la extensión del tumor con una caída distal abrupta.

### Ecuación de Bethe–Bloch

Describe la pérdida media de energía por unidad de longitud de una partícula cargada en un medio:

$$-\frac{dE}{dx} = K\,z^2\,\frac{Z}{A}\,\frac{1}{\beta^2}\left[\frac{1}{2}\ln\frac{2 m_e c^2 \beta^2 \gamma^2 T_{\max}}{I^2} - \beta^2\right].$$

**Identificación de los símbolos:**

- $K = 4\pi N_A r_e^2 m_e c^2 \approx 0.307\ \mathrm{MeV\,cm^2/mol}$: constante.
- $z$: carga del proyectil (en unidades de $e$); para el protón, $z=1$.
- $Z, A$: número atómico y masa atómica del medio (juntos, $Z/A$, dan la densidad de electrones por unidad de masa).
- $\beta = v/c$: velocidad reducida del proyectil; $\gamma = 1/\sqrt{1-\beta^2}$: factor de Lorentz.
- $m_e c^2 = 0.511\ \text{MeV}$: energía en reposo del electrón.
- $T_{\max}$: energía cinética máxima transferible a un electrón libre en una sola colisión.
- $I$: **potencial medio de excitación** del medio, constante característica del material que parametriza la energía promedio transferida a los electrones atómicos (para agua, $I \approx 75\ \text{eV}$).

**Origen del factor $z^2$.** En una colisión de Coulomb, el impulso transversal que el proyectil transfiere a un electrón es proporcional a su carga, $\Delta p \propto z$. La energía transferida es $\Delta E = \Delta p^2/(2m_e) \propto z^2$. Por tanto, la pérdida de energía escala con el **cuadrado de la carga** del proyectil.

**Origen del factor $1/\beta^2$.** El impulso transferido también depende del tiempo de interacción, que es inversamente proporcional a la velocidad ($\Delta p \propto 1/v$). En consecuencia $\Delta E \propto 1/v^2$, es decir $-dE/dx \propto 1/\beta^2$. Este factor implica que **la pérdida de energía aumenta a medida que el proyectil se frena**, y es el responsable directo del pico de Bragg.

(El logaritmo proviene de integrar la energía transferida sobre todos los parámetros de impacto posibles; el cociente dentro del log compara la máxima transferencia $T_{\max}$ con la energía de ligadura típica $I$.)

### Energy straggling

La pérdida de energía es un proceso **estadístico**: resulta de un número enorme de colisiones individuales aleatorias. Por ello, tras recorrer un espesor $\Delta x$, la energía perdida fluctúa en torno a su valor medio con una varianza dada, en la aproximación gaussiana de Bohr, por

$$\sigma_E^2 = 4\pi r_e^2\,(m_e c^2)^2\,N_e\,\frac{z^2}{\beta^2}\,\Delta x,$$

donde $r_e$ es el radio clásico del electrón, $N_e$ la densidad electrónica del medio, $z$ la carga del proyectil, $\beta=v/c$ y $\Delta x$ el espesor recorrido.

**Consecuencia.** Esta fluctuación produce el *range straggling*: protones con la misma energía inicial se detienen a profundidades ligeramente distintas. Esto **ensancha y limita la altura** del pico de Bragg, suavizando la singularidad ideal del modelo a un máximo finito.

### Dispersión Coulombiana múltiple

Al atravesar el medio, el protón sufre muchas pequeñas desviaciones angulares por colisiones con los núcleos atómicos. El ángulo de dispersión RMS acumulado se describe, en la aproximación de Highland, por

$$\theta_0 = \frac{13.6\ \text{MeV}}{\beta c\, p}\,z\,\sqrt{\frac{\Delta x}{X_0}}\left[1 + 0.038\,\ln\!\left(\frac{\Delta x}{X_0}\right)\right],$$

donde $\theta_0$ es el ancho RMS de la distribución angular (aproximadamente gaussiana), $p$ y $\beta c$ son el momento y la velocidad del proyectil, $z$ su carga, $\Delta x$ el espesor atravesado y $X_0$ la **longitud de radiación** del medio (para agua, $X_0 \approx 36\ \text{cm}$). El término logarítmico es una corrección empírica.

**Consecuencia.** La dispersión angular se traduce en un **ensanchamiento lateral del haz** (la penumbra), que degrada la nitidez del borde lateral de la distribución de dosis y debe tenerse en cuenta en la planificación del tratamiento. Nótese que $\theta_0 \propto 1/(\beta c\,p)$: los protones más energéticos se dispersan menos.

# Inciso B)

```python
import numpy as np
from scipy.integrate import quad


K     = 0.307075   # MeV cm^2/mol  (constante de Bethe-Bloch)
z     = 1          # carga del proton (en unidades de e)
Z_A   = 0.55509    # Z/A efectivo del agua (mol/g): (2*1 + 8)/(2*1.008 + 16.00)
me_c2 = 0.510999   # MeV  (energia en reposo del electron)
Mp_c2 = 938.272    # MeV  (energia en reposo del proton)
I     = 75e-6      # MeV  (potencial medio de excitacion del agua = 75 eV)
me_M  = me_c2 / Mp_c2

def beta_gamma(T):
    # A partir de la energia cinetica T del proton (relativista):
    gamma = 1 + T/Mp_c2          # factor de Lorentz
    beta2 = 1 - 1/gamma**2       # beta^2 = (v/c)^2
    return beta2, gamma
```

```python
def stopping_power(T):
    """Poder de frenado masico electronico en MeV cm^2/g, segun Bethe-Bloch."""
    beta2, gamma = beta_gamma(T)
    # Energia maxima transferible a un electron en una colision
    Tmax = 2*me_c2*beta2*gamma**2 / (1 + 2*gamma*me_M + me_M**2)
    # Termino entre corchetes
    L = 0.5*np.log(2*me_c2*beta2*gamma**2*Tmax / I**2) - beta2
    return K * z**2 * Z_A / beta2 * L
```

```python
def csda_range(E0, Emin=1.0):
    """Integral de dE / S(E) desde Emin hasta E0. Resultado en g/cm^2 (= cm en agua)."""
    integrando = lambda E: 1.0 / stopping_power(E)
    R, _ = quad(integrando, Emin, E0, limit=200)
    return R
```

```python
print(f"{'E0 (MeV)':>10}{'S (MeV cm2/g)':>18}{'R_CSDA (g/cm2)':>18}")
for E0 in [50, 150, 250]:
    S = stopping_power(E0)
    R = csda_range(E0)
    print(f"{E0:>10}{S:>18.4f}{R:>18.3f}")
```

```python
# Valores de referencia de NIST PSTAR (agua líquida):
# columna "Electronic Stp. Pow." y "CSDA Range" en E = 50, 150, 250 MeV
nist = {
    50:  {"S": 12.44, "R": 2.227},
    150: {"S": 5.443, "R": 15.77},
    250: {"S": 3.910, "R": 37.94},
}

print(f"{'E0':>5} | {'S_mod':>8}{'S_NIST':>8}{'err%':>7} | {'R_mod':>8}{'R_NIST':>8}{'err%':>7}")
print("-"*55)
for E0 in [50, 150, 250]:
    S = stopping_power(E0)
    R = csda_range(E0)
    Sn, Rn = nist[E0]["S"], nist[E0]["R"]
    err_S = 100*abs(S - Sn)/Sn
    err_R = 100*abs(R - Rn)/Rn
    print(f"{E0:>5} | {S:>8.3f}{Sn:>8.3f}{err_S:>6.2f}% | {R:>8.3f}{Rn:>8.3f}{err_R:>6.2f}%")
```

### Implementación de Bethe–Bloch y cálculo del rango CSDA

Se implementó la ecuación de Bethe–Bloch para protones en agua ($I = 75$ eV), usando cinemática relativista para obtener $\beta$ y $\gamma$ a partir de la energía cinética. El rango CSDA se obtuvo integrando numéricamente

$$R_\text{CSDA}(E_0) = \int_{E_\text{min}}^{E_0}\frac{dE}{(-dE/dx)(E)},$$

desde un $E_\text{min} = 1$ MeV (el rango por debajo de esa energía es despreciable, $\sim 10^{-3}$ g/cm²).

**Discusión.**

- La concordancia con los valores tabulados de NIST es excelente: el error es inferior al 0.3% en todos los casos, y por debajo de 0.05% en cuatro de los seis. Esto valida la implementación de Bethe–Bloch.

- La comparación se hizo contra el poder de frenado **electrónico** de NIST, que es lo que describe Bethe–Bloch. El frenado nuclear es despreciable en este rango de energías (< 0.1%), de modo que el poder de frenado total es prácticamente igual al electrónico.

- La mayor discrepancia (≈ 0.31%) aparece en el rango CSDA a 50 MeV. Esto es esperable: a energías más bajas el integrando recibe más peso de la región de baja energía, donde (i) el corte $E_\text{min}$ omite una fracción pequeña pero no nula del recorrido, y (ii) las **correcciones de capa** (*shell corrections*), despreciadas en la forma básica de Bethe–Bloch, empiezan a ser relevantes. NIST sí las incluye.

- Físicamente, el poder de frenado **disminuye** al aumentar la energía (dominado por el factor $1/\beta^2$), mientras que el rango **crece fuertemente** con $E_0$. De hecho, los rangos siguen aproximadamente la relación de Bragg–Kleeman $R \propto E_0^{\,1.77}$: por ejemplo, $R(150)/R(50) = 7.08$, muy cercano a $(150/50)^{1.77} = 6.99$, y $R(250)/R(50) = 17.0$ frente a $(250/50)^{1.77} = 17.3$. Esto confirma el comportamiento rango–energía usado para describir la profundidad del pico de Bragg en el inciso a).

# Inciso C)

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la simulación
E0    = 150.0     # MeV  (energía inicial)
N     = 10000     # número de protones
dx    = 0.01      # cm   (= 0.1 mm)
zmax  = 18.0      # cm 
nbins = int(zmax/dx)
rho   = 1.0       # g/cm^3 (agua)

def simular_proton():
    """Transporta UN protón por pasos de dx (CSDA, sin fluctuación).
       Devuelve la energía depositada en cada bin de profundidad."""
    E = E0
    dep = np.zeros(nbins)
    i = 0                              # índice de paso = índice de bin (z = i*dx)
    while E > 0 and i < nbins:
        S  = stopping_power(E) * rho   # MeV/cm
        dE = S * dx                    # energía perdida en este paso
        if dE >= E or S <= 0:          # último paso: deposita lo que queda y se detiene
            dE = E
        dep[i] += dE                   # deposita la energía a esta profundidad
        E -= dE
        i += 1
    return dep

# Acumular sobre los N protones
edep = np.zeros(nbins)
for _ in range(N):
    edep += simular_proton()

# Nota: como NO hay fluctuaciones, todos los protones son idénticos. Por eso esto equivale a 'edep = N * simular_proton()', que corre al instante. Mantenemos el bucle porque en los incisos siguientes (con fluctuaciones) cada protón será distinto.
```

```python
z_centers = (np.arange(nbins) + 0.5) * dx
D = edep / (N * dx)               # energía depositada por protón, por unidad de profundidad (MeV/cm)

i_peak = np.argmax(edep)
z_peak = z_centers[i_peak]
R = csda_range(E0)

print(f"Pico de Bragg en z = {z_peak:.3f} cm")
print(f"R_CSDA (inciso b)  = {R:.3f} cm")
print(f"Energía total depositada = {edep.sum():.0f} MeV   (N*E0 = {N*E0:.0f})")

plt.figure(figsize=(9, 5.5))
plt.plot(z_centers, D, color="tab:blue", lw=1.8)
plt.axvline(R, ls="--", color="tab:red", label=f"$R_{{CSDA}}$ = {R:.2f} cm")
plt.axvline(z_peak, ls=":", color="k", label=f"Pico de Bragg = {z_peak:.2f} cm")
plt.xlabel("Profundidad z (cm)"); plt.ylabel("Energía depositada por protón (MeV/cm)")
plt.title("Curva de dosis en profundidad — protones 150 MeV (CSDA)")
plt.xlim(0, 17); plt.legend(); plt.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

### Simulación del transporte (CSDA) y verificación del pico de Bragg

Se simuló el transporte de N = 10⁴ protones de E₀ = 150 MeV en agua, mediante pasos de Δx = 0.1 mm. En cada paso el protón pierde ΔE = (−dE/dx)·Δx (usando el poder de frenado de Bethe–Bloch del inciso b) y esa energía se deposita a la profundidad correspondiente, hasta que el protón se detiene. La suma sobre todos los protones da la curva de dosis en profundidad D(z).

**Resultados.**

- La curva reproduce la forma característica del pico de Bragg: un *plateau* de baja dosis a la entrada (~5 MeV/cm), un ascenso gradual y un máximo agudo y localizado al final del recorrido, seguido de una caída abrupta a cero.
- El pico se ubica en **z ≈ 15.79 cm**, en excelente acuerdo con el rango calculado en el inciso b, **R_CSDA ≈ 15.76 cm**. La diferencia es del orden de la resolución del bin (0.01 cm), lo que verifica la consistencia entre ambos métodos: en la aproximación CSDA, el protón deposita su máxima energía justo antes de detenerse, es decir, a una profundidad igual a su rango.
- Como verificación adicional, la energía total depositada es exactamente N·E₀ = 1.5×10⁶ MeV, confirmando la conservación de la energía.

**Interpretación.** El pico emerge directamente del factor 1/β² de Bethe–Bloch: al frenarse, el protón deposita cada vez más energía por unidad de camino, concentrando la dosis al final del recorrido. Dado que en esta aproximación no hay fluctuaciones estadísticas, todos los protones se detienen exactamente a la misma profundidad y el pico resulta extremadamente estrecho (idealmente, una singularidad). En un escenario realista, el *energy straggling* hace que los protones se detengan a profundidades ligeramente distintas, ensanchando el pico a una altura y anchura finitas — efecto que se incorporará en el siguiente inciso.

```python
import numpy as np
import matplotlib.pyplot as plt

# Constante de straggling de Bohr para agua
C = K * me_c2 * rho * Z_A      # MeV^2/cm 

E0    = 150.0
N     = 10000
dx    = 0.01                   # cm
zmax  = 18.0
nbins = int(zmax/dx)

def simular(con_straggling, seed=0):
    """Transporta N protones (vectorizado). Devuelve la energía depositada por
       bin y la profundidad de detención de cada protón."""
    rng = np.random.default_rng(seed)
    E       = np.full(N, E0)
    stopped = np.zeros(N, bool)
    z_stop  = np.full(N, np.nan)
    edep    = np.zeros(nbins)

    for i in range(nbins):
        act = ~stopped
        if not act.any():
            break
        Ea = E[act]
        g  = 1 + Ea/Mp_c2
        b2 = 1 - 1/g**2
        mean_dE = stopping_power(Ea) * rho * dx       # pérdida media en el paso
        invalid = mean_dE <= 0                        # Bethe-Bloch ya no vale (E muy baja)

        if con_straggling:
            var  = C/b2 * dx                          # sigma_E^2 del paso
            safe = np.where(invalid, 1.0, mean_dE)    # evita parámetros inválidos
            # gamma con misma media y varianza -> no negativa, sin sesgo
            dE = rng.gamma(safe**2/var, var/safe)
        else:
            dE = mean_dE.copy()

        dE = np.where(invalid, Ea, dE)                # si inválido, deposita el resto
        over = dE >= Ea                               # último paso: no pierde más de lo que tiene
        dE = np.where(over, Ea, dE)

        edep[i] += dE.sum()
        Enew = Ea - dE
        E[act] = Enew
        just = over | (Enew <= 0)                     # protones que se detienen aquí
        idx = np.where(act)[0][just]
        z_stop[idx] = (i+1)*dx
        stopped[idx] = True

    return edep, z_stop
```

```python
edep_sin, _      = simular(con_straggling=False)
edep_con, z_stop = simular(con_straggling=True)

z_centers = (np.arange(nbins) + 0.5) * dx
R = csda_range(E0)
sigma_R = np.nanstd(z_stop)        # ensanchamiento: dispersión de los rangos

print(f"R_CSDA                       = {R:.3f} cm")
print(f"Rango medio (con straggling) = {np.nanmean(z_stop):.3f} cm")
print(f"sigma_R                      = {sigma_R*10:.3f} mm  ({100*sigma_R/R:.2f}% de R)")

D_sin = edep_sin/(N*dx)
D_con = edep_con/(N*dx)

fig, ax = plt.subplots(1, 2, figsize=(14, 5.5))
for a, (lo, hi) in zip(ax, [(0,17), (14,16.5)]):
    a.plot(z_centers, D_sin, color="gray", lw=1.5, ls="--", label="sin straggling (CSDA)")
    a.plot(z_centers, D_con, color="tab:blue", lw=1.8, label="con straggling")
    a.axvline(R, ls=":", color="tab:red", alpha=.7, label=f"$R_{{CSDA}}$={R:.2f} cm")
    a.set_xlabel("Profundidad z (cm)"); a.set_ylabel("Energía depositada por protón (MeV/cm)")
    a.set_xlim(lo, hi); a.grid(alpha=0.3); a.legend()
ax[0].set_title("Curva completa")
ax[1].set_title(f"Zoom al pico ($\\sigma_R$ = {sigma_R*10:.2f} mm)")
plt.tight_layout(); plt.show()
```

### Transporte con straggling y ensanchamiento del pico

Se repitió la simulación del inciso c) incorporando la fluctuación estadística de la pérdida de energía en cada paso, $\Delta E \sim \mathcal{N}(\langle\Delta E\rangle, \sigma_E^2)$, con la varianza de Bohr $\sigma_E^2 = (K\,m_e c^2\,\rho\,Z/A)\,(z^2/\beta^2)\,\Delta x$.

**Nota de implementación.** A alta energía $\sigma_E$ es comparable a $\langle\Delta E\rangle$ por paso, por lo que una gaussiana literal genera valores $\Delta E < 0$ (no físicos). Para evitar el sesgo del rango (al truncar en cero) y los artefactos de dosis negativa, se muestreó $\Delta E$ de una distribución con la misma media y varianza pero no negativa (gamma), que coincide con la gaussiana cuando $\sigma_E \ll \langle\Delta E\rangle$.

**Resultados.**

- Con straggling, el rango medio (15.80 cm) sigue coincidiendo con $R_\text{CSDA}$ (15.76 cm): la fluctuación **no** desplaza la posición media del pico, solo lo ensancha.
- El pico de Bragg se **reduce y se ensancha**: la altura cae de ~150 MeV/cm (caso ideal CSDA, casi singular) a ~27 MeV/cm, y la caída distal deja de ser un acantilado vertical para volverse una pendiente suave de varios milímetros. La región de entrada (*plateau*) no se ve afectada.
- El ensanchamiento medido, definido como la desviación estándar de las profundidades de detención de los protones, es

$$\sigma_R \approx 3.7\ \text{mm} \quad (\approx 2.3\%\ \text{de }R).$$

la sobreestimación proviene de que la forma simple de Bohr empleada incluye el factor $1/\beta^2$, mientras que el straggling real es casi independiente de la velocidad.

**Relevancia clínica.** La principal ventaja terapéutica del protón es la **caída distal abrupta** del pico de Bragg, que permite no irradiar el tejido sano situado detrás del tumor. El straggling degrada justamente esa caída: introduce una **penumbra distal** de unos milímetros, de modo que parte del tejido inmediatamente posterior al tumor recibe dosis. En consecuencia, la posición del borde distal del tratamiento tiene una **incertidumbre de rango** del orden de $\sigma_R$. Por eso, en la planificación de tratamientos se añaden **márgenes distales** (típicamente ~3.5% del rango más unos milímetros), y se evita en lo posible "apuntar" el borde distal del haz directamente contra un órgano crítico (médula espinal, nervio óptico, etc.), ya que un error de pocos milímetros puede significar sobredosis en ese órgano o subdosis en el tumor.
