### Introducción a la Fase 2: Discretización y Marco Numérico

En esta segunda fase del proyecto, el objetivo principal es construir el puente entre el modelo matemático teórico (desarrollado en la Fase 1) y su futura implementación computacional. Partiendo de la ecuación diferencial de tipo Helmholtz que gobierna el sistema unidimensional, el propósito es "discretizar" el dominio espacial continuo, transformándolo en una cuadrícula finita de puntos evaluables por un computador. 

Para garantizar una alta fidelidad en la simulación de las plumas de inestabilidad, se aplicarán esquemas de diferencias finitas compactas de alto orden (hasta 7 puntos). A lo largo de esta sección, traduciremos las ecuaciones diferenciales y sus condiciones de borde a un lenguaje algebraico discreto, con el fin de ensamblar un sistema matricial estructurado que represente las interacciones físicas de todo el fluido.

**(a): Discretización para $p=1$ y $p=J-1$**

Partimos de la ecuación de Helmholtz unidimensional:
$$
\frac{\partial^2 u}{\partial x^2} = \theta(x)\,u(x) - \psi(x)
$$

En cada punto de la malla $x_p = p\Delta x$, con $p=0,1,\dots,J$, se tiene:

$$\
u''_p = \theta_p u_p - \psi_p \qquad \text{(1)}
$$ 
donde 
$$\theta_p=\theta(x_p), \psi_p=\psi(x_p)$$

El esquema de diferencias finitas compactas de cuarto orden (Padé) para $p=1$ (y análogo para $p=J-1$) es:

$$
\frac{1}{10} u''_{p-1} + u''_p + \frac{1}{10} u''_{p+1} = \frac{12}{10} \left( \frac{u_{p+1} - 2u_p + u_{p-1}}{\Delta x^2} \right) \qquad \text{(2)}
$$

**Formación de terminos para $u''_{p-1}, u''_p, u''_{p+1}$** \
A través de la relación (1), obtenemos para:
$$\ u''_{p-1}=\theta_{p-1}u_{p-1} - \psi_{p-1} \qquad \text{(1.1)}$$
$$\ u''_p=\theta_p u_p - \psi_p \qquad \text{(1.2)}$$
$$\ u''_{p+1}=\theta_{p+1}u_{p+1} - \psi_{p+1} \qquad \text{(1.3)}$$
Los términos anteriores[(1.1),(1.2) y (1.3)], procedemos a sustituirlos según corresponda en la ecuación (2):
$$
\frac{1}{10}(\theta_{p-1}u_{p-1} - \psi_{p-1}) + (\theta_p u_p - \psi_p) + \frac{1}{10}(\theta_{p+1}u_{p+1} - \psi_{p+1})
= \frac{12}{10\Delta x^2}(u_{p+1} - 2u_p + u_{p-1}) \qquad \text{(2.1)}
$$

Continuamos,expandiendo del lado derecho de (2.1):

$$
\frac{12}{10\Delta x^2}u_{p+1} - \frac{24}{10\Delta x^2}u_p + \frac{12}{10\Delta x^2}u_{p-1} \qquad \text{(2.2)}
$$

Luego, restamos el lado derecho completo al lado izquierdo y reordenamos (2.2), de forma que todos los términos que contienen $u$ al lado izquierdo y los términos con $\psi$ al lado derecho. 
$$\frac{1}{10}\theta_{p-1}u_{p-1} + \theta_p u_p + \frac{1}{10}\theta_{p+1}u_{p+1}
- \frac{12}{10\Delta x^2}u_{p+1} + \frac{24}{10\Delta x^2}u_p - \frac{12}{10\Delta x^2}u_{p-1}
= \frac{1}{10}\psi_{p-1} + \psi_p + \frac{1}{10}\psi_{p+1} \qquad \text{(2.3)}$$ 

Agrupamos los coeficientes para $u_{p-1}$, $u_p$ y $u_{p+1}$:
 $$ (\frac{1}{10}\theta_{p-1} - \frac{12}{10\Delta x^2})u_{p-1} +(\theta_p + \frac{12}{5\Delta x^2})u_p + (\frac{1}{10}\theta_{p+1} - \frac{12}{10\Delta x^2})u_{p+1}= \frac{1}{10}\psi_{p-1} + \psi_p + \frac{1}{10}\psi_{p+1} \qquad \text{(3)}$$

**Recordemos la ecuación (21)del enunciado**: $$c_p^{(1)} u_{p-1} + d_p^{(1)} u_p + e_p^{(1)} u_{p+1} = h_p^{(1)}, \qquad p = 1,\;J-1$$

**Identificar los coeficientes de la ecuación discretizada para p = 1 y p= J-1**\
Finalmente, utilizando la ecuación (3) relacionamos sus terminos con los de la ecuación (21) de enunciado:
$$ 
\begin{aligned}
h_p^{(1)} &= \frac{1}{10}\psi_{p-1} + \psi_p + \frac{1}{10}\psi_{p+1} \\[2mm]
c_p^{(1)} &= \frac{1}{10}\theta_{p-1} - \frac{12}{10\Delta x^2} \\[2mm]
d_p^{(1)} &= \theta_p + \frac{12}{5\Delta x^2} \\[2mm]
e_p^{(1)} &= \frac{1}{10}\theta_{p+1} - \frac{12}{10\Delta x^2}
\end{aligned}
$$

**Nota final:** Para $p=1$, nos aparece multiplicando al coeficiente $c_p^{(1)}$, el término $u_0$ que se reemplaza mediante la condición de borde (19) y para $p=J-1$ sucede algo similar con el coeficiente $e_p^{(1)}$, donde aparece el término $u_J$ que se reemplaza con la ecuación (20).A

**(b): Discretización para $p=2$ y $p=J-2$**

Volvemos a la ecuación de Helmholtz:
$$
\frac{\partial^2 u}{\partial x^2} = \theta(x)\,u(x) - \psi(x)
$$
Y a la ecuación (1) del item anterior.

El esquema de diferencias finitas compactas de cuarto orden para $p=2$ (y análogo para $p=J-2$) siendo la ecuación (17) del enunciado:
$$
\tilde\beta\, u''_{p-2} + \tilde\alpha\, u''_{p-1} + u''_p + \tilde\alpha\, u''_{p+1} + \tilde\beta\, u''_{p+2}
= \tilde b\,\frac{u_{p+2} - 2u_p + u_{p-2}}{4\Delta x^2}
+ \tilde a\,\frac{u_{p+1} - 2u_p + u_{p-1}}{\Delta x^2} \qquad \text{(4)}
$$

con los coeficientes conocidos:

$$
\tilde\alpha = 0.40437617,\quad \tilde\beta = 0.02221463,\quad
\tilde a = 0.49797006,\quad \tilde b = 1.35521152.
$$
**Formación de terminos para $u''_{p-2}, u''_{p-1}, u''_p, u''_{p+1}, u''_{p+2}$** \
A través de la relación (1) tenemos: 
$$
\ u''_{p-2} = \theta_{p-2}u_{p-2} - \psi_{p-2} \qquad$$
$$\ u''_{p-1}= \theta_{p-1}u_{p-1} - \psi_{p-1} \qquad$$
$$\ u''_p = \theta_p u_p - \psi_p \qquad$$
$$\ u''_{p+1} = \theta_{p+1}u_{p+1} - \psi_{p+1} \qquad$$
$$\ u''_{p+2} = \theta_{p+2}u_{p+2} - \psi_{p+2}$$
Procedemos a sustituir estos terminos en la ecuación (4):
$$
\begin{aligned}
&\tilde\beta(\theta_{p-2}u_{p-2} - \psi_{p-2})
+ \tilde\alpha(\theta_{p-1}u_{p-1} - \psi_{p-1})
+ (\theta_p u_p - \psi_p) \\
&+ \tilde\alpha(\theta_{p+1}u_{p+1} - \psi_{p+1})
+ \tilde\beta(\theta_{p+2}u_{p+2} - \psi_{p+2}) \\
&= \frac{\tilde b}{4\Delta x^2}(u_{p+2} - 2u_p + u_{p-2})
+ \frac{\tilde a}{\Delta x^2}(u_{p+1} - 2u_p + u_{p-1})
\end{aligned}
\qquad \text{(4.1)}
$$

Continuamos, expandiendo el lado derecho de la ecuación (4.1):

$$
\frac{\tilde b}{4\Delta x^2}u_{p+2} - \frac{\tilde b}{2\Delta x^2}u_p + \frac{\tilde b}{4\Delta x^2}u_{p-2}
+ \frac{\tilde a}{\Delta x^2}u_{p+1} - \frac{2\tilde a}{\Delta x^2}u_p + \frac{\tilde a}{\Delta x^2}u_{p-1} \qquad\text{(4.2)}
$$

Restamos los terminos del lado derecho (4.2) y reordenamos todos los términos, de forma que los que contienen $u$ van al lado izquierdo y los términos con $\psi$ van al lado derecho. 

$$
\begin{aligned}
&\tilde\beta\theta_{p-2}u_{p-2} + \tilde\alpha\theta_{p-1}u_{p-1} + \theta_p u_p + \tilde\alpha\theta_{p+1}u_{p+1} + \tilde\beta\theta_{p+2}u_{p+2} \\
&- \left( \frac{\tilde b}{4\Delta x^2}u_{p+2} - \frac{\tilde b}{2\Delta x^2}u_p + \frac{\tilde b}{4\Delta x^2}u_{p-2} \right) \\
&- \left( \frac{\tilde a}{\Delta x^2}u_{p+1} - \frac{2\tilde a}{\Delta x^2}u_p + \frac{\tilde a}{\Delta x^2}u_{p-1} \right) \\
&= \tilde\beta\psi_{p-2} + \tilde\alpha\psi_{p-1} + \psi_p + \tilde\alpha\psi_{p+1} + \tilde\beta\psi_{p+2} 
\end{aligned} \qquad\text{(4.3)}
$$

Agruparmos los coeficientes para cada $u_{p-2}$,$u_{p-1}$,$u_p$,$u_{p+1}$ y $u_{p+2}$:
$$
\begin{aligned}
 &(\tilde\beta\theta_{p-2} - \frac{\tilde b}{4\Delta x^2})u_{p-2} + (\tilde\alpha\theta_{p-1} - \frac{\tilde a}{\Delta x^2})u_{p-1} +\\ 
&(\theta_p + \frac{\tilde b}{2\Delta x^2} + \frac{2\tilde a}{\Delta x^2})u_p +(\tilde\alpha\theta_{p+1} - \frac{\tilde a}{\Delta x^2})u_{p+1}+\\
&(\tilde\beta\theta_{p+2} - \frac{\tilde b}{4\Delta x^2})u_{p+2} = \tilde\beta\psi_{p-2} + \tilde\alpha\psi_{p-1} + \psi_p + \tilde\alpha\psi_{p+1}+ \tilde\beta\psi_{p+2}
\end{aligned} \qquad\text{(5)}
$$

**Recordamos la ecuación discretizada (22) del enunciado:** 
$$
b_p^{(2)} u_{p-2} + c_p^{(2)} u_{p-1} + d_p^{(2)} u_p + e_p^{(2)} u_{p+1} + f_p^{(2)} u_{p+2} = h_p^{(2)}, \qquad p = 2,\;J-2
$$
**Identificar los coeficientes de la ecuación discretizada para p = 1 y p= J-1**\
Finalmente, usando la ecuación (5) relacionamos sus terminos con los de la ecuación (22):
$$
\begin{aligned}
b_p^{(2)} &= \tilde\beta\theta_{p-2} - \frac{\tilde b}{4\Delta x^2} \\[2mm]
c_p^{(2)} &= \tilde\alpha\theta_{p-1} - \frac{\tilde a}{\Delta x^2} \\[2mm]
d_p^{(2)} &= \theta_p + \frac{\tilde b}{2\Delta x^2} + \frac{2\tilde a}{\Delta x^2} \\[2mm]
e_p^{(2)} &= \tilde\alpha\theta_{p+1} - \frac{\tilde a}{\Delta x^2} \\[2mm]
f_p^{(2)} &= \tilde\beta\theta_{p+2} - \frac{\tilde b}{4\Delta x^2} \\[2mm]
h_p^{(2)} &= \tilde\beta\psi_{p-2} + \tilde\alpha\psi_{p-1} + \psi_p + \tilde\alpha\psi_{p+1} + \tilde\beta\psi_{p+2}
\end{aligned}
$$

**Nota final:** En símil del caso de (a) tenemos que para $p=2$, aparece $u_0$ en $b_2^{(2)}$; para $p=J-2$, aparece $u_J$ en $f_{J-2}^{(2)}$. Estos valores se reemplazan utilizando las condiciones de borde discretizadas (19) y (20).B

### c) Método Híbrido: Discretización Septadiagonal (Nodos Interiores)

**1. Sustitución de la Ecuación Modelo**

Para los nodos interiores $p = 3, 4, \dots, J-3$, el esquema de diferencias finitas compactas está dado por la Ecuación:
$$
\gamma u''_{p-3} + \beta u''_{p-2} + \alpha u''_{p-1} + u''_p + \alpha u''_{p+1} + \beta u''_{p+2} + \gamma u''_{p+3} = c \left[ \frac{u_{p+3} - 2u_p + u_{p-3}}{9\Delta x^2} \right] + b \left[ \frac{u_{p+2} - 2u_p + u_{p-2}}{4\Delta x^2} \right] + a \left[ \frac{u_{p+1} - 2u_p + u_{p-1}}{\Delta x^2} \right]
$$

El modelo generalizado de la Ecuación que debemos reemplazar ahí establece que la segunda derivada en cualquier nodo $k$ es:
$$
u''_k = \theta_k u_k - \psi_k
$$

Procedemos a reemplazar esta definición en cada uno de los términos del lado izquierdo de la Ecuación:
$$
\gamma(\theta_{p-3} u_{p-3} - \psi_{p-3}) + \beta(\theta_{p-2} u_{p-2} - \psi_{p-2}) + \alpha(\theta_{p-1} u_{p-1} - \psi_{p-1}) + (\theta_p u_p - \psi_p) + \alpha(\theta_{p+1} u_{p+1} - \psi_{p+1}) + \beta(\theta_{p+2} u_{p+2} - \psi_{p+2}) + \gamma(\theta_{p+3} u_{p+3} - \psi_{p+3})
$$
$$
= \frac{c}{9\Delta x^2} u_{p+3} - \frac{2c}{9\Delta x^2} u_p + \frac{c}{9\Delta x^2} u_{p-3} + \frac{b}{4\Delta x^2} u_{p+2} - \frac{2b}{4\Delta x^2} u_p + \frac{b}{4\Delta x^2} u_{p-2} + \frac{a}{\Delta x^2} u_{p+1} - \frac{2a}{\Delta x^2} u_p + \frac{a}{\Delta x^2} u_{p-1}
$$

**2. Agrupación de Términos (Construcción del Sistema Lineal)**

Para formar el sistema algebraico, debemos trasladar todas las incógnitas (los términos $u_k$) hacia el lado izquierdo de la igualdad, y agrupar todos los términos fuente conocidos (los términos $\psi_k$) hacia el lado derecho.

Agrupando cuidadosamente las variables $u_k$ desde $p-3$ hasta $p+3$:

* **Para $u_{p-3}$:** $\quad \left( \gamma \theta_{p-3} - \frac{c}{9\Delta x^2} \right) u_{p-3}$
* **Para $u_{p-2}$:** $\quad \left( \beta \theta_{p-2} - \frac{b}{4\Delta x^2} \right) u_{p-2}$
* **Para $u_{p-1}$:** $\quad \left( \alpha \theta_{p-1} - \frac{a}{\Delta x^2} \right) u_{p-1}$
* **Para $u_p$:** Al pasar los términos negativos del lado derecho al izquierdo, se suman. Notando que $\frac{2b}{4} = \frac{b}{2}$, obtenemos:
    $$\left( \theta_p + \frac{2c}{9\Delta x^2} + \frac{b}{2\Delta x^2} + \frac{2a}{\Delta x^2} \right) u_p$$
* **Para $u_{p+1}$:** $\quad \left( \alpha \theta_{p+1} - \frac{a}{\Delta x^2} \right) u_{p+1}$
* **Para $u_{p+2}$:** $\quad \left( \beta \theta_{p+2} - \frac{b}{4\Delta x^2} \right) u_{p+2}$
* **Para $u_{p+3}$:** $\quad \left( \gamma \theta_{p+3} - \frac{c}{9\Delta x^2} \right) u_{p+3}$

Para el lado derecho, enviamos todos los términos con $-\psi_k$, los cuales cambian de signo a positivo.

**3. Identificación de Coeficientes Finales**

Al comparar nuestra agrupación con la estructura deseada:
$$a_p^{(3)} u_{p-3} + b_p^{(3)} u_{p-2} + c_p^{(3)} u_{p-1} + d_p^{(3)} u_p + e_p^{(3)} u_{p+1} + f_p^{(3)} u_{p+2} + g_p^{(3)} u_{p+3} = h_p^{(3)}$$

Se deducen los siguientes coeficientes analíticos para la matriz septadiagonal:

$$\boxed{ a_p^{(3)} = \gamma \theta_{p-3} - \frac{c}{9(\Delta x)^2} }$$
$$\boxed{ b_p^{(3)} = \beta \theta_{p-2} - \frac{b}{4(\Delta x)^2} }$$
$$\boxed{ c_p^{(3)} = \alpha \theta_{p-1} - \frac{a}{(\Delta x)^2} }$$
$$\boxed{ d_p^{(3)} = \theta_p + \frac{2c}{9(\Delta x)^2} + \frac{b}{2(\Delta x)^2} + \frac{2a}{(\Delta x)^2} }$$
$$\boxed{ e_p^{(3)} = \alpha \theta_{p+1} - \frac{a}{(\Delta x)^2} }$$
$$\boxed{ f_p^{(3)} = \beta \theta_{p+2} - \frac{b}{4(\Delta x)^2} }$$
$$\boxed{ g_p^{(3)} = \gamma \theta_{p+3} - \frac{c}{9(\Delta x)^2} }$$


Finalmente, el término fuente modificado que reside en el vector solución derecho es:
$$\boxed{ h_p^{(3)} = \gamma \psi_{p-3} + \beta \psi_{p-2} + \alpha \psi_{p-1} + \psi_p + \alpha \psi_{p+1} + \beta \psi_{p+2} + \gamma \psi_{p+3} }$$

### d) Ensamblaje del Sistema Lineal Septadiagonal $\bar{P}\mathbf{u} = \mathbf{H}$

Para construir la matriz $\bar{P}$ de dimensiones $(J-1) \times (J-1)$, debemos mapear los nodos físicos $p \in [1, J-1]$ a las filas de la matriz indexadas como $i \in [0, J-2]$. La regla general de correspondencia es que la fila $i$ de la matriz representa la ecuación evaluada en el nodo físico $p = i+1$. 

Dado que el vector de incógnitas es $\mathbf{u} = [u_1, u_2, \dots, u_{J-1}]^T$, los valores "fantasmas" de los bordes $u_0$ y $u_J$ no pertenecen al sistema y deben ser eliminados sustituyendo las siguientes condiciones de borde en los nodos adyacentes:
$$u_0 = \bar{a}^{(1)} - \bar{b}^{(1)} u_1 + \bar{c}^{(1)} u_2 - \bar{d}^{(1)} u_3$$
$$u_J = \bar{a}^{(2)} + \bar{b}^{(2)} u_{J-1} - \bar{c}^{(2)} u_{J-2} + \bar{d}^{(2)} u_{J-3}$$

---

**1. Efecto del Borde Inferior ($u_0$) en las Filas Superiores**

Sustituimos la expresión de $u_0$ en las primeras tres filas físicas que dependen de él:

* **Fila 0 ($p=1$):** Utilizando la Ecuación del inciso (A), se tiene $c_1^{(1)} u_0 + d_1^{(1)} u_1 + e_1^{(1)} u_2 = h_1^{(1)}$. Al sustituir $u_0$ y agrupar:
  * $D_0 = d_1^{(1)} - c_1^{(1)}\bar{b}^{(1)}$
  * $E_0 = e_1^{(1)} + c_1^{(1)}\bar{c}^{(1)}$
  * $F_0 = -c_1^{(1)}\bar{d}^{(1)}$
  * $H_0 = h_1^{(1)} - c_1^{(1)}\bar{a}^{(1)}$

* **Fila 1 ($p=2$):** Utilizando la Ecuación del inciso (B), $b_2^{(2)} u_0 + c_2^{(2)} u_1 + d_2^{(2)} u_2 + e_2^{(2)} u_3 + f_2^{(2)} u_4 = h_2^{(2)}$. Al sustituir $u_0$ y agrupar:
  * $C_1 = c_2^{(2)} - b_2^{(2)}\bar{b}^{(1)}$
  * $D_1 = d_2^{(2)} + b_2^{(2)}\bar{c}^{(1)}$
  * $E_1 = e_2^{(2)} - b_2^{(2)}\bar{d}^{(1)}$
  * $F_1 = f_2^{(2)}$
  * $H_1 = h_2^{(2)} - b_2^{(2)}\bar{a}^{(1)}$

* **Fila 2 ($p=3$):** Utilizando la Ecuación del inciso (C) de 7 puntos. Sustituyendo $u_0$ en el término $a_3^{(3)}u_0$:
  * $B_2 = b_3^{(3)} - a_3^{(3)}\bar{b}^{(1)}$
  * $C_2 = c_3^{(3)} + a_3^{(3)}\bar{c}^{(1)}$
  * $D_2 = d_3^{(3)} - a_3^{(3)}\bar{d}^{(1)}$
  * $E_2 = e_3^{(3)}$
  * $F_2 = f_3^{(3)}$
  * $G_2 = g_3^{(3)}$
  * $H_2 = h_3^{(3)} - a_3^{(3)}\bar{a}^{(1)}$

---

**2. Nodos Interiores Centrales**

Para las filas lo suficientemente alejadas de los bordes, donde la molécula de 7 puntos de la Ecuación del inciso (C) no toca ni $u_0$ ni $u_J$. Esto corresponde a los nodos físicos $p=4, \dots, J-4$ (es decir, las filas $i = 3, \dots, J-5$). Los coeficientes se transfieren directamente:

* $A_i = a_{i+1}^{(3)}$
* $B_i = b_{i+1}^{(3)}$
* $C_i = c_{i+1}^{(3)}$
* $D_i = d_{i+1}^{(3)}$
* $E_i = e_{i+1}^{(3)}$
* $F_i = f_{i+1}^{(3)}$
* $G_i = g_{i+1}^{(3)}$
* $H_i = h_{i+1}^{(3)}$

---

**3. Efecto del Borde Superior ($u_J$) en las Filas Inferiores**

Sustituimos la Ecuación (20), $u_J = \bar{a}^{(2)} + \bar{b}^{(2)} u_{J-1} - \bar{c}^{(2)} u_{J-2} + \bar{d}^{(2)} u_{J-3}$, en las últimas tres filas:

* **Fila $J-4$ ($p=J-3$):** Usando la Ecuación del inciso (C) de 7 puntos, el término $g_{J-3}^{(3)} u_J$ se expande:
  * $A_{J-4} = a_{J-3}^{(3)}$
  * $B_{J-4} = b_{J-3}^{(3)}$
  * $C_{J-4} = c_{J-3}^{(3)}$
  * $D_{J-4} = d_{J-3}^{(3)} + g_{J-3}^{(3)}\bar{d}^{(2)}$
  * $E_{J-4} = e_{J-3}^{(3)} - g_{J-3}^{(3)}\bar{c}^{(2)}$
  * $F_{J-4} = f_{J-3}^{(3)} + g_{J-3}^{(3)}\bar{b}^{(2)}$
  * $H_{J-4} = h_{J-3}^{(3)} - g_{J-3}^{(3)}\bar{a}^{(2)}$

* **Fila $J-3$ ($p=J-2$):** Usando la Ecuación del inciso (B), el término $f_{J-2}^{(2)} u_J$ se expande:
  * $B_{J-3} = b_{J-2}^{(2)}$
  * $C_{J-3} = c_{J-2}^{(2)} + f_{J-2}^{(2)}\bar{d}^{(2)}$
  * $D_{J-3} = d_{J-2}^{(2)} - f_{J-2}^{(2)}\bar{c}^{(2)}$
  * $E_{J-3} = e_{J-2}^{(2)} + f_{J-2}^{(2)}\bar{b}^{(2)}$
  * $H_{J-3} = h_{J-2}^{(2)} - f_{J-2}^{(2)}\bar{a}^{(2)}$

* **Fila $J-2$ ($p=J-1$):** Usando la Ecuación del inciso (A), el término $e_{J-1}^{(1)} u_J$ se expande:
  * $B_{J-2} = e_{J-1}^{(1)}\bar{d}^{(2)}$
  * $C_{J-2} = c_{J-1}^{(1)} - e_{J-1}^{(1)}\bar{c}^{(2)}$
  * $D_{J-2} = d_{J-1}^{(1)} + e_{J-1}^{(1)}\bar{b}^{(2)}$
  * $H_{J-2} = h_{J-1}^{(1)} - e_{J-1}^{(1)}\bar{a}^{(2)}$

---

**4. Representación Matricial Final del Sistema**

Con todos los coeficientes identificados y los valores de contorno asimilados en las filas extremas, el sistema global de ecuaciones lineales toma la forma matricial explícita $\bar{P}\mathbf{u} = \mathbf{H}$:

$$
\begin{bmatrix} 
D_0 & E_0 & F_0 & 0 & 0 & 0 & \dots & 0 & 0 & 0 & 0 & 0 & 0 \\
C_1 & D_1 & E_1 & F_1 & 0 & 0 & \dots & 0 & 0 & 0 & 0 & 0 & 0 \\
B_2 & C_2 & D_2 & E_2 & F_2 & G_2 & \dots & 0 & 0 & 0 & 0 & 0 & 0 \\
A_3 & B_3 & C_3 & D_3 & E_3 & F_3 & G_3 & \dots & 0 & 0 & 0 & 0 & 0 \\
0 & A_4 & B_4 & C_4 & D_4 & E_4 & F_4 & \dots & 0 & 0 & 0 & 0 & 0 \\
\vdots & \ddots & \ddots & \ddots & \ddots & \ddots & \ddots & \ddots & \ddots & \vdots & \vdots & \vdots & \vdots \\
0 & 0 & 0 & 0 & 0 & \dots & A_{J-5} & B_{J-5} & C_{J-5} & D_{J-5} & E_{J-5} & F_{J-5} & G_{J-5} \\
