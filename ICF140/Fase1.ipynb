## 1. Introducción
El presente proyecto estudia el transporte de masa en 2D de una mezcla de fluidos miscibles en un medio poroso. El fenómeno principal es la inestabilidad de **Rayleigh-Taylor**, la cual genera plumas convectivas que mejoran la transferencia de masa. Este estudio es relevante para la dinámica de solubilidad del $CO_2$ en acuíferos, una estrategia clave contra el cambio climático.



## 2. Marco Teórico y Desarrollo Matemático:
Las ecuaciones que gobiernan este problema de dinámica en un medio poroso (aproximación de Boussinesq) son:

* (1a) **Incompresibilidad:** $\nabla^* \cdot u^* = 0$ 
* (1b) **Ecuación de Darcy:** $u^* \frac{\mu}{K} = -\nabla^* p^* - \rho^*(c) g \hat{z}$ 
* (1c) **Advección-Difusión:** $\phi \frac{\partial c}{\partial t^*} + \nabla^* \cdot (c u^*) = \phi \mathcal{D} \nabla^{*2} c$ 
* (1d) **Ecuación de Estado:** $\rho^*(c) = \rho_0 + \Delta \rho c$ 

Donde para abordar las ecuaciones anterior y comenzar a desarrollar, se busca adimensionarla con el siguiente esquema de cambio de variables:
* $x^* = Hx$, $z^* = Hz$
* $u^* = u_c u$
* $p^* = \beta p$
* $t^* = \tau t$

donde $H$ es la altura del dominio (longitud característica), y $u_c$, $\beta$, $\tau$ son escalas dimensionales a determinar. La concentración $c$ ya es adimensional ($0\le c\le1$).

Comenzamos con la ecuación (1a) $\nabla^* \cdot u^* = 0$, donde el gradiente dimensional $\nabla^*$ se relaciona con el gradiente adimensional $\nabla$, tomando en consideración que $\mathbf{x}^* = H\mathbf{x}$,por cada derivada espacial aparece un factor $1/H$
$$
\frac{\partial}{\partial x^*} = \frac{1}{H}\frac{\partial}{\partial x}
$$
Con lo anterior en mente, se puede utilizar la siguiente relación del gradiente:
$$
\nabla^* = \frac{1}{H}\nabla
$$
Donde al sustituir el otro factor de la ecuación (1a) $\mathbf{u}^* = u_c\mathbf{u}$, nos queda:
$$
\nabla^*\cdot\mathbf{u}^* = \frac{1}{H}\nabla\cdot(u_c\mathbf{u}) = \frac{u_c}{H}\nabla\cdot\mathbf{u} = 0
$$
Y como $u_c/H\neq0$, se obtiene la **ecuación de incompresibilidad adimensional**:
$$
\boxed{\nabla\cdot\mathbf{u}=0}
$$

Continuamos con la ecuación de Darcy (1b)  $u^* \frac{\mu}{K} = -\nabla^* p^* - \rho^*(c) g \hat{z}$ , se procede a extender en todo sus terminos utilizando la ecuación de estado (1d) $\rho^*(c) = \rho_0 + \Delta \rho c$ en la expresión (1b):
$$
\mu K^{-1}\mathbf{u}^* = -\nabla^*p^* - \rho_0 g\,\hat{\mathbf{z}} - \Delta\rho\,c\,g\,\hat{\mathbf{z}} 
$$
Y se reemplazan estos factores $\mathbf{u}^* = u_c\mathbf{u}$, $p^* = \beta p$, $\nabla^* = \frac{1}{H}\nabla$, llegando a la expresión extendida:
$$
\mu K^{-1}u_c\mathbf{u} = -\frac{\beta}{H}\nabla p - \rho_0 g\,\hat{\mathbf{z}} - \Delta\rho\,c\,g\,\hat{\mathbf{z}} 
$$
Se observa que el término constante $-\rho_0 g\,\hat{\mathbf{z}}$ puede escribirse como un gradiente:
$$
-\rho_0 g\,\hat{\mathbf{z}} = -\nabla^*\bigl(\rho_0 g z^*\bigr) = -\frac{1}{H}\nabla\bigl(\rho_0 g H z\bigr) = -\nabla\bigl(\rho_0 g z\bigr)
$$
Se usa $z^* = H z$, por lo tanto, absorbemos ese término en una presión modificada, definiendo una nueva presión adimensional $\tilde{p}$ tal que:
$$
p^* = \beta\,\tilde{p} - \rho_0 g z^*
$$
o equivalentemente,
$$
\beta\,\tilde{p} = p^* + \rho_0 g z^*.
$$
Entonces $\nabla^*p^* = \nabla^*(\beta\tilde{p} - \rho_0 g z^*) = \beta\nabla^*\tilde{p} - \rho_0 g\hat{\mathbf{z}}$.

Sustituyendo esta relación en la ecuación de darcy extendida:
$$
\mu K^{-1}u_c\mathbf{u} = -\beta\nabla^*\tilde{p} + \rho_0 g\hat{\mathbf{z}} - \rho_0 g\hat{\mathbf{z}} - \Delta\rho\,c\,g\,\hat{\mathbf{z}} = -\beta\nabla^*\tilde{p} - \Delta\rho\,c\,g\,\hat{\mathbf{z}}.
$$

Ahora adimensionalizamos $\nabla^*\tilde{p} = \frac{1}{H}\nabla\tilde{p}$, y obtenemos:
$$
\mu K^{-1}u_c\mathbf{u} = -\frac{\beta}{H}\nabla\tilde{p} - \Delta\rho\,c\,g\,\hat{\mathbf{z}} \tag{4.3}
$$

Dividimos ambos lados por $\mu u_c/K$:
$$
\mathbf{u} = -\frac{\beta K}{\mu u_c H}\,\nabla\tilde{p} \;-\; \frac{\Delta\rho\,g\,K}{\mu u_c}\,c\,\hat{\mathbf{z}} \tag{4.4}
$$

Elegimos las escalas para simplificar:
- Hacemos el coeficiente del término de flotabilidad igual a $1$:
  $$
  \frac{\Delta\rho\,g\,K}{\mu u_c}=1 \quad\Longrightarrow\quad \boxed{u_c = \frac{\Delta\rho\,g\,K}{\mu}}
  $$
- Hacemos el coeficiente del gradiente de presión igual a $1$ (para obtener la forma más sencilla):
  $$
  \frac{\beta}{\Delta\rho\,g\,H}=1 \quad\Longrightarrow\quad \boxed{\beta = \Delta\rho\,g\,H}
  $$

Con estos valores, la ecuación se reduce a:
$$
\boxed{\mathbf{u} = -\nabla\tilde{p} - c\,\hat{\mathbf{z}}}
$$

Renombramos $\tilde{p}$ como $p$ (presión dinámica adimensional). Así la **ecuación de Darcy adimensional** final es:
$$
\boxed{\mathbf{u} = -\nabla p - c\,\hat{\mathbf{z}}}
$$

Proseguimos con la ecuación de advección-difusión (1c) $\phi \frac{\partial c}{\partial t^*} + \nabla^* \cdot (c u^*) = \phi \mathcal{D} \nabla^{*2} c$, en donde consideramos la ecuación dimensional con $\phi=1$, consiguiendo la expresión de a continuación:
$$
\frac{\partial c}{\partial t^*} + \nabla^*\cdot(c\mathbf{u}^*) = D\,\nabla^{*2}c
$$
Donde, se procede a transformar cada término correspondiente:
- $\displaystyle \frac{\partial}{\partial t^*} = \frac{1}{\tau}\frac{\partial}{\partial t}$
- $\displaystyle \nabla^*\cdot(c\mathbf{u}^*) = \frac{1}{H}\nabla\cdot(c\,u_c\mathbf{u}) = \frac{u_c}{H}\,\nabla\cdot(c\mathbf{u})$
- Usando $\nabla\cdot\mathbf{u}=0$, $\nabla\cdot(c\mathbf{u}) = \mathbf{u}\cdot\nabla c$
- $\displaystyle \nabla^{*2}c = \frac{1}{H^2}\nabla^2 c$

Para sustituirlos en la ecuación y obtener:
$$
\frac{1}{\tau}\frac{\partial c}{\partial t} + \frac{u_c}{H}\,\mathbf{u}\cdot\nabla c = \frac{D}{H^2}\nabla^2 c
$$

Luego, se multiplica por $\tau$:
$$
\frac{\partial c}{\partial t} + \frac{\tau u_c}{H}\,\mathbf{u}\cdot\nabla c = \frac{\tau D}{H^2}\nabla^2 c \tag{5.1}
$$

Y se considera para que el coeficiente del término convectivo tenga un valor de $1$, la siguiente igualdad:
$$
\frac{\tau u_c}{H}=1 \quad\Longrightarrow\quad \boxed{\tau = \frac{H}{u_c}}
$$

Se vuelve a sustituir $\tau$ en el coeficiente difusivo, llegando a:
$$
\frac{\tau D}{H^2} = \frac{H}{u_c}\cdot\frac{D}{H^2} = \frac{D}{u_c H}
$$

Finalmente, se define el **número de Rayleigh** como:
$$
\boxed{Ra = \frac{u_c H}{D}}
$$
Entonces la expresión final para la **ecuación de advección-difusión** queda:
$$
\boxed{\frac{\partial c}{\partial t} + \mathbf{u}\cdot\nabla c = \frac{1}{Ra}\,\nabla^2 c}
$$
**Nota**: 
El número de Rayleigh compara la rapidez del transporte por convección (movimiento del fluido) frente al transporte por difusión (movimiento molecular), donde tenemos la escala de tiempo convectiva $\tau = \frac{H}{u_c}$(tiempo que tarda una partícula en recorrer la altura $H$ con velocidad $u_c$) y la escala de tiempo difusiva  $\tau_d = \frac{H^2}{\mathcal{D}}$(tiempo característico para que la difusión molecular actúe en la escala $H$). Obteniendo una relacion entre ambos tiempos se consigue que $Ra$ es proporcional a la razón $\frac{\tau_d}{\tau}$. Considerando lo anterior,  a valores de $Ra \gg 1$: $\tau_d \gg \tau$ → la convección domina sobre la difusión y valores de $Ra \ll 1$: $\tau_d \ll \tau$ → la difusión es muy rápida y tiende a homogeneizar.

Finalmente, las escalas dimensionales encontradas fueron:
$$
\boxed{
u_c = \frac{\Delta\rho\,g\,K}{\mu}, \qquad
\beta = \Delta\rho\,g\,H, \qquad
\tau = \frac{H}{u_c} = \frac{\mu H}{\Delta\rho\,g\,K}
}
$$




### b) Formulación Corriente-Vorticidad y Reducción a la Ecuación de Poisson

**1. Demostración de la incompresibilidad mediante la función de corriente**

A partir de los resultados del inciso anterior, la condición de incompresibilidad en 2D se expresa como:
$$
\nabla \cdot \mathbf{u} = \frac{\partial u}{\partial x} + \frac{\partial w}{\partial z} = 0
$$

Se propone el siguiente cambio de variables basado en la función de corriente escalar $\psi(x,z)$:
$$
u = -\frac{\partial \psi}{\partial z} \quad ; \quad w = \frac{\partial \psi}{\partial x}
$$

Sustituyendo estas definiciones en la ecuación de incompresibilidad:
$$
\frac{\partial}{\partial x}\left(-\frac{\partial \psi}{\partial z}\right) + \frac{\partial}{\partial z}\left(\frac{\partial \psi}{\partial x}\right) = -\frac{\partial^2 \psi}{\partial x \partial z} + \frac{\partial^2 \psi}{\partial z \partial x}
$$

Por el Teorema de Clairaut (simetría de las segundas derivadas parciales cruzadas para funciones continuas), se cumple que $\frac{\partial^2 \psi}{\partial x \partial z} = \frac{\partial^2 \psi}{\partial z \partial x}$. Por lo tanto:
$$
-\frac{\partial^2 \psi}{\partial x \partial z} + \frac{\partial^2 \psi}{\partial x \partial z} = 0
$$

Queda demostrado que la formulación mediante la función de corriente **satisface idéntica e inmediatamente la ecuación de incompresibilidad**.

---

**2. Eliminación de la presión y deducción de la Ecuación de Poisson**

Partimos de la ecuación de Darcy adimensional obtenida en el inciso a):
$$
\mathbf{u} = -\nabla p - c\,\hat{\mathbf{z}}
$$

Separando la ecuación vectorial en sus componentes cartesianas $x$ y $z$:
$$
u = -\frac{\partial p}{\partial x} \tag{B.1}
$$
$$
w = -\frac{\partial p}{\partial z} - c \tag{B.2}
$$

Para eliminar el término de la presión, aplicamos una derivación cruzada (equivalente a aplicar el operador rotacional $\nabla \times$). Derivamos la componente $u$ respecto a $z$, y la componente $w$ respecto a $x$:
$$
\frac{\partial u}{\partial z} = -\frac{\partial^2 p}{\partial x \partial z}
$$
$$
\frac{\partial w}{\partial x} = -\frac{\partial^2 p}{\partial z \partial x} - \frac{\partial c}{\partial x}
$$

Restando la segunda expresión de la primera $\left(\frac{\partial u}{\partial z} - \frac{\partial w}{\partial x}\right)$ para cancelar los términos de presión cruzada:
$$
\frac{\partial u}{\partial z} - \frac{\partial w}{\partial x} = \left(-\frac{\partial^2 p}{\partial x \partial z}\right) - \left(-\frac{\partial^2 p}{\partial z \partial x} - \frac{\partial c}{\partial x}\right)
$$
$$
\frac{\partial u}{\partial z} - \frac{\partial w}{\partial x} = \frac{\partial c}{\partial x} \tag{B.3}
$$

A continuación, reescribimos el lado izquierdo de la ecuación (B.3) en términos de la función de corriente $\psi$:
$$
\frac{\partial u}{\partial z} = \frac{\partial}{\partial z}\left(-\frac{\partial \psi}{\partial z}\right) = -\frac{\partial^2 \psi}{\partial z^2}
$$
$$
\frac{\partial w}{\partial x} = \frac{\partial}{\partial x}\left(\frac{\partial \psi}{\partial x}\right) = \frac{\partial^2 \psi}{\partial x^2}
$$

Sustituyendo esto en (B.3):
$$
-\frac{\partial^2 \psi}{\partial z^2} - \frac{\partial^2 \psi}{\partial x^2} = \frac{\partial c}{\partial x}
$$
$$
-\left( \frac{\partial^2 \psi}{\partial x^2} + \frac{\partial^2 \psi}{\partial z^2} \right) = \frac{\partial c}{\partial x}
$$

Reconociendo que el término en paréntesis corresponde al operador Laplaciano en 2D ($\nabla^2 \psi$), llegamos a la ecuación final:
$$
\boxed{\nabla^2 \psi = -\frac{\partial c}{\partial x}}
$$

---

**3. Determinación de la función de vorticidad $\omega(x,z)$**

El problema solicita demostrar que la ecuación toma la forma $\nabla^2 \psi = -\omega(x, z)$. Al comparar esta estructura con nuestro resultado empírico de la sección anterior, se determina directamente que la **función de vorticidad adimensional** en términos de la concentración es:
$$
\boxed{\omega(x, z) = \frac{\partial c}{\partial x}}
$$

*(Adicionalmente, expresada en términos del campo cinemático, la vorticidad de un flujo en 2D se define como $\omega = \frac{\partial u}{\partial z} - \frac{\partial w}{\partial x} = -\nabla^2 \psi$).*

---

**4. Condiciones de borde para la función de corriente**

Las condiciones de borde físicas de no penetración en el dominio cerrado rectangular son:
* Paredes laterales ($x=0$ y $x=L$): $u = 0$
* Techo y fondo ($z=0$ y $z=H$): $w = 0$

Transformando estas condiciones para $\psi$:
1.  En $x=0$ y $x=L$, $u = -\frac{\partial \psi}{\partial z} = 0$. Esto implica que $\psi$ no varía respecto a $z$; por ende, es una constante a lo largo de las paredes verticales.
2.  En $z=0$ y $z=H$, $w = \frac{\partial \psi}{\partial x} = 0$. Esto implica que $\psi$ no varía respecto a $x$; por ende, es una constante a lo largo de las paredes horizontales.

Como todo el contorno físico del dominio está interconectado, la función $\psi$ debe poseer el mismo valor constante en toda la frontera (todo el borde actúa como una única línea de corriente). Dado que las velocidades se calculan a partir de derivadas espaciales de $\psi$, el valor numérico de esta constante es arbitrario. Por simplicidad matemática, se define como cero.

Por lo tanto, las condiciones de borde para la función de corriente son de Dirichlet homogéneo:
$$
\boxed{\psi(x,z) = 0 \quad \text{para el contorno } \partial\Omega}
$$

### c) Método Híbrido: Derivación de la ODE tipo Helmholtz

**1. Sustitución de las expansiones en la Ecuación de Poisson**

Se desea resolver la ecuación elíptica de Poisson en 2D:
$$
\nabla^2 \psi = \frac{\partial^2 \psi}{\partial x^2} + \frac{\partial^2 \psi}{\partial z^2} = -\omega(x,z)
$$

El método propone expandir tanto la función de corriente $\psi$ como la vorticidad $\omega$ en series de senos a lo largo del eje $x$, garantizando automáticamente el cumplimiento de las condiciones de Dirichlet en $x=0$ y $x=L$:
$$
\psi(x,z) = \sum_{k=0}^{\infty} F_k(z) \sin\left(\frac{k\pi x}{L}\right) \tag{C.1}
$$
$$
\omega(x,z) = \sum_{k=0}^{\infty} G_k(z) \sin\left(\frac{k\pi x}{L}\right) \tag{C.2}
$$

Procedemos a calcular las segundas derivadas espaciales de la función $\psi$. Como la sumatoria es una operación lineal, las derivadas parciales actúan únicamente sobre los términos correspondientes:

* **Derivada respecto a $z$:** La variable $x$ se comporta como constante, por lo que la derivada recae exclusivamente sobre la amplitud $F_k(z)$:
    $$
    \frac{\partial^2 \psi}{\partial z^2} = \sum_{k=0}^{\infty} \frac{d^2 F_k(z)}{dz^2} \sin\left(\frac{k\pi x}{L}\right)
    $$

* **Derivada respecto a $x$:** La amplitud $F_k(z)$ se comporta como constante, y derivamos la función seno dos veces, aplicando la regla de la cadena:
    $$
    \frac{\partial \psi}{\partial x} = \sum_{k=0}^{\infty} F_k(z) \left(\frac{k\pi}{L}\right) \cos\left(\frac{k\pi x}{L}\right)
    $$
    $$
    \frac{\partial^2 \psi}{\partial x^2} = -\sum_{k=0}^{\infty} F_k(z) \left(\frac{k\pi}{L}\right)^2 \sin\left(\frac{k\pi x}{L}\right)
    $$

**2. Obtención de la Ecuación de Helmholtz**

Sustituimos estas segundas derivadas y la expansión de $\omega(x,z)$ en la ecuación de Poisson original:
$$
\sum_{k=0}^{\infty} \left[ \frac{d^2 F_k(z)}{dz^2} - \left(\frac{k\pi}{L}\right)^2 F_k(z) \right] \sin\left(\frac{k\pi x}{L}\right) = -\sum_{k=0}^{\infty} G_k(z) \sin\left(\frac{k\pi x}{L}\right)
$$

Debido a que las funciones $\sin\left(\frac{k\pi x}{L}\right)$ forman un conjunto ortogonal en el intervalo $[0, L]$, la única forma de que esta igualdad se cumpla para todo $x$ es que los coeficientes que acompañan al seno de cada modo $k$ sean idénticos a ambos lados de la ecuación. 

Igualando los términos dentro de los corchetes, los sumatorios desaparecen y obtenemos una **Ecuación Diferencial Ordinaria (ODE)** para cada modo $k$:
$$
\boxed{ \frac{d^2 F_k(z)}{dz^2} - \left(\frac{k\pi}{L}\right)^2 F_k(z) = -G_k(z) }
$$

Esta es la clásica **Ecuación de Helmholtz 1D no homogénea**, donde $-\left(\frac{k\pi}{L}\right)^2$ representa el autovalor del operador Laplaciano en el dominio transversal.

---

**3. Condiciones de Borde para las amplitudes $F_k(z)$**

El enunciado establece condiciones de borde mixtas generales en la dirección $z$:
$$
\alpha_0 \psi(x,0) + \beta_0 \left.\frac{\partial \psi}{\partial z}\right|_{z=0} = \gamma_1(x) \tag{C.3}
$$
$$
\alpha_1 \psi(x,1) + \beta_1 \left.\frac{\partial \psi}{\partial z}\right|_{z=1} = \gamma_2(x) \tag{C.4}
$$

Para que el sistema sea consistente, las funciones fuente en los bordes, $\gamma_1(x)$ y $\gamma_2(x)$, también deben expandirse en series de Fourier de senos:
$$
\gamma_1(x) = \sum_{k=0}^{\infty} C_{1,k} \sin\left(\frac{k\pi x}{L}\right) \quad ; \quad \gamma_2(x) = \sum_{k=0}^{\infty} C_{2,k} \sin\left(\frac{k\pi x}{L}\right)
$$
Donde las constantes se determinan mediante la transformada: $C_{i,k} = \frac{2}{L}\int_0^L \gamma_i(x)\sin\left(\frac{k\pi x}{L}\right) dx$.

Sustituyendo la expansión de $\psi$ (Ecuación C.1) en la condición inferior (C.3):
$$
\alpha_0 \sum_{k=0}^{\infty} F_k(0) \sin\left(\frac{k\pi x}{L}\right) + \beta_0 \sum_{k=0}^{\infty} \left.\frac{dF_k}{dz}\right|_{z=0} \sin\left(\frac{k\pi x}{L}\right) = \sum_{k=0}^{\infty} C_{1,k} \sin\left(\frac{k\pi x}{L}\right)
$$

Agrupando términos y apelando nuevamente a la ortogonalidad, obtenemos las condiciones de borde ordinarias para la amplitud $F_k(z)$:

**Condición en $z=0$:**
$$
\alpha_0 F_k(0) + \beta_0 \left.\frac{dF_k}{dz}\right|_{z=0} = C_{1,k}
$$

**Condición en $z=1$:**
$$
\alpha_1 F_k(1) + \beta_1 \left.\frac{dF_k}{dz}\right|_{z=1} = C_{2,k}
$$

*Nota de consistencia física:* Para nuestro problema específico deducido en la sección b), la función de corriente es cero en todo el contorno (Dirichlet homogéneo). Esto implica que $\alpha_0=\alpha_1=1$, $\beta_0=\beta_1=0$, y $\gamma_1(x)=\gamma_2(x)=0$. Por lo tanto, para la simulación de Rayleigh-Taylor, las condiciones de borde de la ODE se reducen simplemente a:
$$
\boxed{ F_k(0) = 0 \quad \text{y} \quad F_k(1) = 0 }
$$


### 4. Conclusiones de la Fase 1

El desarrollo de esta primera fase ha permitido establecer un marco teórico robusto y computacionalmente eficiente para el estudio de la inestabilidad de Rayleigh-Taylor en medios porosos. A partir del modelamiento matemático, se destacan los siguientes hitos fundamentales:

1. **Caracterización Física y Adimensionalización:** Mediante el escalamiento de las ecuaciones gobernantes de Darcy y Advección-Difusión, se eliminó la dependencia de parámetros físicos aislados, demostrando que la dinámica del sistema está controlada fundamentalmente por el **Número de Rayleigh solutal ($Ra$)**. Este parámetro adimensional se reveló como el indicador clave que dictamina si el transporte de masa estará dominado por la difusión molecular (homogeneización lenta) o por la convección (plumas gravitacionales rápidas).

2. **Reducción del Sistema Acoplado:** La implementación de la formulación corriente-vorticidad ($\psi - \omega$) demostró ser una herramienta analítica invaluable. Al definir el campo de velocidades a través de una función de corriente escalar, se garantizó el cumplimiento intrínseco de la incompresibilidad y se logró aislar y eliminar el campo de presiones termodinámicas, reduciendo la ecuación vectorial de Darcy a una única ecuación elíptica escalar de Poisson ($\nabla^2 \psi = -\omega$).

3. **Viabilidad Numérica mediante el Método Híbrido:** La mayor victoria analítica de esta fase fue la aplicación de métodos espectrales. Aprovechando las condiciones de borde de Dirichlet homogéneas en el eje horizontal, la expansión en series de senos de Fourier permitió desacoplar la dependencia espacial. Esto transformó una Ecuación Diferencial Parcial (EDP) bidimensional compleja en una familia de Ecuaciones Diferenciales Ordinarias (EDO) independientes de tipo Helmholtz, gobernadas por el número de onda $k$.

**Perspectivas para la Fase 2:**
El sistema físico complejo original ha sido traducido con éxito a un modelo matemático linealizado en una de sus dimensiones. Este marco híbrido sienta las bases perfectas para la discretización numérica en la siguiente etapa, donde las EDOs unidimensionales resultantes podrán ser resueltas eficientemente mediante métodos estándar de diferencias finitas en el eje vertical $z$, garantizando alta precisión espectral y bajos costos computacionales.
