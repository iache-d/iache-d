# Demostración Analítica: Problema Lineal Asociado

Para realizar el análisis de estabilidad lineal, se asume que el fluido se encuentra en un estado base de difusión pura, sobre el cual se introducen perturbaciones infinitesimales modeladas como ondas planas. 

El ansatz es:
* Velocidad horizontal: $u(t, x, z) = U(t, z) e^{ikx}$
* Velocidad vertical: $w(t, x, z) = W(t, z) e^{ikx}$
* Concentración: $c(t, x, z) = c_b(t, z) + c'(t, z) e^{ikx}$

Donde $c_b(t, z)$ satisface la ecuación de difusión del estado base (Ec. 16):
$$\frac{\partial c_b}{\partial t} = \frac{1}{Ra} \frac{\partial^2 c_b}{\partial z^2}$$

El principio fundamental de la **linealización** consiste en despreciar los productos entre perturbaciones (términos proporcionales a $e^{2ikx}$, $e^{3ikx}$, etc.), ya que al ser perturbaciones minúsculas, su producto tiende a cero.

---

### Parte 1: Deducción de la Ecuación tipo Helmholtz

Partimos de la ecuación de Poisson para la función de corriente y la definición de velocidad vertical:
$$\nabla^2 \psi = -\frac{\partial c}{\partial x} \quad \text{y} \quad w = \frac{\partial \psi}{\partial x}$$

**1. Formulación en términos de $w$:**
Aplicamos el operador derivada parcial respecto a $x$ a ambos lados de la ecuación de Poisson. Dado que los operadores espaciales conmutan:
$$\frac{\partial}{\partial x} \left( \nabla^2 \psi \right) = \nabla^2 \left( \frac{\partial \psi}{\partial x} \right) = -\frac{\partial^2 c}{\partial x^2}$$
Sustituyendo la definición de $w$:
$$\nabla^2 w = -\frac{\partial^2 c}{\partial x^2}$$

**2. Expansión y sustitución del Ansatz:**
Expandiendo el Laplaciano en 2D ($x, z$):
$$\frac{\partial^2 w}{\partial x^2} + \frac{\partial^2 w}{\partial z^2} = -\frac{\partial^2 c}{\partial x^2}$$

Calculamos las segundas derivadas respecto a $x$ utilizando nuestras soluciones de ondas planas (recordando que $i^2 = -1$):
* $\frac{\partial^2 w}{\partial x^2} = -k^2 W e^{ikx}$
* $\frac{\partial^2 c}{\partial x^2} = -k^2 c' e^{ikx}$ (notar que $\frac{\partial c_b}{\partial x} = 0$)

Sustituyendo en la ecuación diferencial:
$$-k^2 W e^{ikx} + \frac{\partial^2 W}{\partial z^2} e^{ikx} = k^2 c' e^{ikx}$$

**3. Simplificación final:**
Dividiendo toda la expresión por el factor común $e^{ikx}$ y reordenando, obtenemos:
$$\frac{\partial^2 W}{\partial z^2} - k^2 W = k^2 c'$$

---

### Parte 2: Deducción de la Ecuación de Transporte Linealizada

Partimos de la ecuación de transporte acoplada:
$$\frac{\partial c}{\partial t} + u \frac{\partial c}{\partial x} + w \frac{\partial c}{\partial z} = \frac{1}{Ra} \nabla^2 c$$

**1. Evaluación de los términos convectivos:**
Sustituimos el ansatz en los términos no lineales de la izquierda:
* **Eje X:** $$u \frac{\partial c}{\partial x} = (U e^{ikx}) \left( ik c' e^{ikx} \right) = ik U c' e^{2ikx} \approx 0$$
  *(Se desprecia por ser de orden superior $\mathcal{O}(e^{2ikx})$).*

* **Eje Z:** $$w \frac{\partial c}{\partial z} = (W e^{ikx}) \left( \frac{\partial c_b}{\partial z} + \frac{\partial c'}{\partial z} e^{ikx} \right) = W \frac{\partial c_b}{\partial z} e^{ikx} + W \frac{\partial c'}{\partial z} e^{2ikx} \approx W \frac{\partial c_b}{\partial z} e^{ikx}$$
  *(Nuevamente, el término no lineal se anula).*

**2. Sustitución en la EDP completa:**
Reemplazamos la expansión en la derivada temporal y en el Laplaciano:
$$\left( \frac{\partial c_b}{\partial t} + \frac{\partial c'}{\partial t} e^{ikx} \right) + W \frac{\partial c_b}{\partial z} e^{ikx} = \frac{1}{Ra} \left[ \frac{\partial^2 c_b}{\partial z^2} + \left( \frac{\partial^2 c'}{\partial z^2} - k^2 c' \right) e^{ikx} \right]$$

**3. Sustracción del Estado Base:**
Reagrupamos los términos que no dependen de la perturbación ($e^{ikx}$):
$$\left[ \frac{\partial c_b}{\partial t} - \frac{1}{Ra} \frac{\partial^2 c_b}{\partial z^2} \right] + \left[ \frac{\partial c'}{\partial t} + W \frac{\partial c_b}{\partial z} \right] e^{ikx} = \frac{1}{Ra} \left( \frac{\partial^2 c'}{\partial z^2} - k^2 c' \right) e^{ikx}$$

Por definición de nuestro estado base, el primer corchete es idénticamente cero. La ecuación se reduce netamente al balance de las perturbaciones:
$$\left( \frac{\partial c'}{\partial t} + W \frac{\partial c_b}{\partial z} \right) e^{ikx} = \frac{1}{Ra} \left( \frac{\partial^2 c'}{\partial z^2} - k^2 c' \right) e^{ikx}$$

**4. Simplificación final:**
Dividiendo por $e^{ikx}$ y despejando la derivada temporal de la perturbación de concentración, llegamos a:
$$\frac{\partial c'}{\partial t} = -W \frac{\partial c_b}{\partial z} + \frac{1}{Ra} \left( \frac{\partial^2 c'}{\partial z^2} - k^2 c' \right)$$
